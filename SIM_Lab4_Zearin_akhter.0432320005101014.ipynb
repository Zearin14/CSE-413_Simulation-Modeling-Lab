```python
import random
import numpy as np
import math
##Setup & personalization
student_id = input("Enter your Student ID: ")
full_name = input("Enter your full name: ")

R = int(student_id[-2:])

number_of_letters = sum(char.isalpha() for char in full_name)

minimum_points = 200 * number_of_letters

radius = R / 10
def is_inside_circle(x, y, radius):
    return (x ** 2 + y ** 2) <= radius ** 2
##simulation logic
def simulate(total_points, radius):

    inside_count = 0
    outside_count = 0

    half_points = total_points // 2
    for _ in range(half_points):

        x = random.uniform(0, radius)
        y = random.uniform(0, radius)

        if is_inside_circle(x, y, radius):
            inside_count += 1
        else:
            outside_count += 1
    for _ in range(total_points - half_points):

        x = np.random.rand() * radius
        y = np.random.rand() * radius

        if is_inside_circle(x, y, radius):
            inside_count += 1
        else:
            outside_count += 1

    estimated_pi = 4 * inside_count / total_points

    error = abs(estimated_pi - math.pi)

    return inside_count, outside_count, estimated_pi, error

print("Monte Carlo Circle Simulation")
print(f"Student ID              : {student_id}")
print(f"Last two digits (R)     : {R}")
print(f"Full Name               : {full_name}")
print(f"Number of letters       : {number_of_letters}")
print(f"Circle radius (R/10)    : {radius}")
print(f"Minimum points          : {minimum_points}")

print(
    f"{'Range Used':<15}"
    f"{'Total Points':<15}"
    f"{'Points Inside':<18}"
    f"{'Estimated π':<18}"
    f"{'Error from math.pi':<20}"
)


for multiplier in range(1, 6):

    total_points = minimum_points * multiplier

    inside, outside, estimated_pi, error = simulate(
        total_points,
        radius
    )

    range_used = f"[0, {radius:.2f}]"

    print(
        f"{range_used:<15}"
        f"{total_points:<15}"
        f"{inside:<18}"
        f"{estimated_pi:<18.6f}"
        f"{error:<20.6f}"
    )

print(f"Actual value of π = {math.pi:.6f}")
```
