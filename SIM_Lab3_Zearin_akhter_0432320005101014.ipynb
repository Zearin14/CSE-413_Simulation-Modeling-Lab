```python
import numpy as np
import matplotlib.pyplot as plt
n=3000
regular_days=np.random.uniform(2,12,n)
busy_days=np.random.normal(8,2,n)
busy_days[busy_days<0]=0
```

2

```python
plt.hist(regular_days, bins=30, alpha=0.5, color='green', edgecolor='red')

plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.title('Histogram of Wait Times - Regular Days')
plt.grid(True)
plt.show()
plt.hist(busy_days, bins=30, alpha=0.5, color='purple', edgecolor='blue')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.title('Histogram of Wait Times - Busy Days')
plt.grid(True)
plt.show()
```

3

```python
print("REGULAR DAYS")
print("Mean:", np.mean(regular_days))
print("Median:", np.median(regular_days))
print("Standard Deviation:", np.std(regular_days))
print("Maximum Wait Time:", np.max(regular_days))
print("\nBUSY DAYS")
print("Mean:", np.mean(busy_days))
print("Median:", np.median(busy_days))
print("Standard Deviation:", np.std(busy_days))
print("Maximum Wait Time:", np.max(busy_days))
```
